## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

This starter slice has enough signal to be interesting. I am focusing on visible pages with real demand and position data because that is where a CTR or engagement opportunity model can help editors sort their review queue.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
from pathlib import Path
import pandas as pd

candidate_paths = [
    Path('data/raw/content_refresh_anonymized.csv'),
    Path.cwd() / 'data' / 'raw' / 'content_refresh_anonymized.csv',
    Path.cwd().parent / 'data' / 'raw' / 'content_refresh_anonymized.csv',
    Path.cwd().parent.parent / 'data' / 'raw' / 'content_refresh_anonymized.csv',
]
data_path = next((p for p in candidate_paths if p.exists()), None)
if data_path is None:
    raise FileNotFoundError('Could not locate data/raw/content_refresh_anonymized.csv')
df = pd.read_csv(data_path)
rows = len(df)
usable_position_rows = int((df['avg_position'] > 0).sum())
high_volume_low_ctr = int(((df['impressions_90d'] >= 500) & (df['avg_position'] > 0) & (df['avg_position'] <= 20) & (df['ctr'] < 0.5)).sum())
high_sessions_weak_engagement = int(((df['sessions_90d'] >= 30) & ((df['engagement_rate'] < 30) | (df['scroll_rate'] < 30))).sum())
print(f'Total rows in starter dataset: {rows}')
print(f'Visible pages with usable position data: {usable_position_rows}')
print(f'High-volume low-CTR pages (>=500 impressions, pos 1-20, CTR < 0.5): {high_volume_low_ctr}')
print(f'High-session pages with weak engagement: {high_sessions_weak_engagement}')
```

This is observational work: strong for prioritization, not proof of causality or a claim about Google’s black box.
